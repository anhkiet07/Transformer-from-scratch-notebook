```python
import sys, os
sys.path.append(os.path.abspath(".."))

from src.Transformer import Transformer
import numpy as np
```

```python
d_model = 512
num_heads = 8
drop_prob = 0.1
batch_size = 3
max_sequence_length = 10
ffn_hidden = 2048
num_layers = 2

START_TOKEN = '<START>'
END_TOKEN = '<END>'
PADDING_TOKEN = '<PADDING>'

vocabulary = [START_TOKEN, END_TOKEN, PADDING_TOKEN] + list("abcdefghijklmnopqrstuvwxyz ")
language_to_index = {v: k for k, v in enumerate(vocabulary)}
```

```python
# Test full Transformer: encoder + decoder + final linear projection
kn_vocab_size = len(vocabulary)

transformer = Transformer(d_model, ffn_hidden, num_heads, drop_prob, num_layers,
                           max_sequence_length, kn_vocab_size,
                           language_to_index, language_to_index,
                           START_TOKEN, END_TOKEN, PADDING_TOKEN)

x_sentences = ["hello", "world", "test"]
y_sentences = ["hi there", "foo bar", "sample"]

out = transformer.forward(x_sentences, y_sentences,
                           enc_start_token=True, enc_end_token=True,
                           dec_start_token=True, dec_end_token=True)
print("Transformer output shape:", out.shape)
assert out.shape == (batch_size, max_sequence_length, kn_vocab_size)
print("OK")
```
