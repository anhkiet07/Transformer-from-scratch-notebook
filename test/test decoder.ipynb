```python
import sys, os
sys.path.append(os.path.abspath(".."))

from src.Transformer import Decoder
import numpy as np
```

```python
d_model = 512
num_heads = 8
drop_prob = 0.1
batch_size = 3
max_sequence_length = 10
ffn_hidden = 2048
num_layers = 2

START_TOKEN = '<START>'
END_TOKEN = '<END>'
PADDING_TOKEN = '<PADDING>'

vocabulary = [START_TOKEN, END_TOKEN, PADDING_TOKEN] + list("abcdefghijklmnopqrstuvwxyz ")
language_to_index = {v: k for k, v in enumerate(vocabulary)}
```

```python
# Test Decoder alone: fake encoder output + a batch of target sentences
decoder = Decoder(d_model, ffn_hidden, num_heads, drop_prob, num_layers,
                   max_sequence_length, language_to_index, START_TOKEN, END_TOKEN, PADDING_TOKEN)

encoder_output = np.random.randn(batch_size, max_sequence_length, d_model)
y = ["hello", "world", "test"]

decoder_out = decoder.forward(encoder_output, y,
                               self_attention_mask=None, cross_attention_mask=None,
                               start_token=True, end_token=True)
print("Decoder output shape:", decoder_out.shape)
assert decoder_out.shape == (batch_size, max_sequence_length, d_model)
print("OK")
```
